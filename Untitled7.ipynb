```python
import pandas as pd

df = pd.read_csv(
    filename,
    encoding="utf-8",
    sep=",",
    dtype={
        "Employee_ID": "string",
        "First_Name": "string",
        "Last_Name": "string",
        "Department_Region": "string",
        "Status": "string",
        "Email": "string",
        "Phone": "string",
        "Performance_Score": "string"
    }
)

print(df.dtypes)
```

```python
print("Performance Scores:")
print(df["Performance_Score"].value_counts(dropna=False))

print("\nStatus values:")
print(df["Status"].unique())

print("\nRemote Work values:")
print(df["Remote_Work"].unique())

print("\nDepartment and Region values:")
print(df["Department_Region"].unique())

print("\nJoin Date examples:")
print(df["Join_Date"].head(15))

print("\nPhone examples:")
print(df["Phone"].head(15))
```

```python
# Convert Join_Date into datetime format

df["Join_Date"] = pd.to_datetime(
    df["Join_Date"],
    format="mixed",
    errors="coerce"
)

print(df["Join_Date"].head(15))

print("\nInvalid dates:")
print(df["Join_Date"].isnull().sum())
```

```python
# Remove unwanted spaces

df["Status"] = df["Status"].str.strip().str.capitalize()

df["Department_Region"] = (
    df["Department_Region"]
    .str.strip()
)

df["Performance_Score"] = (
    df["Performance_Score"]
    .str.strip()
)

# Check results

print(df["Status"].unique())

print(df["Department_Region"].unique())

print(df["Performance_Score"].unique())
```

```python

# Step 6: Handle Missing Values and Correct Phone Numbers

# Fill missing Age values with median
df['Age'] = df['Age'].fillna(df['Age'].median())

# Fill missing Salary values with median
df['Salary'] = df['Salary'].fillna(df['Salary'].median())

# Correct negative phone numbers
df['Phone'] = df['Phone'].astype(str).str.replace('-', '', regex=False)

# Check missing values
print("Missing Values:")
print(df.isnull().sum())

# Check phone numbers
print("\nCorrected Phone Numbers:")
print(df['Phone'].head(15))
```

```python
# Export cleaned dataset
df.to_csv("cleaned_employee_data.csv", index=False)

# Download the cleaned file
from google.colab import files
files.download("cleaned_employee_data.csv")
```
